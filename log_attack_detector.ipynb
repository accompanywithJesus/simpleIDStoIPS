```python
#replace to definition
#함수를 건드려야 할 것 같은데
result1= '192.168.0.1 - - [01/Dec/2025:20:55:36 +0900] "GET /admin/login.php HTTP/1.1" 404 2326'

def parse_log(line):
  part = line.split()
  ip = part[0]
  time = "".join(part[3:5]).strip('[]')
  attack_url = part[5]
  return ip,time,attack_url#어쨋든 return을 해야 함
#부족한 개념 함수의 scope.return 뒤로의 함수는 죽은 코드임.
```

```python
  #12.06. Today's goal is Mass Processing and pattern detecting
log_list = [
    '192.168.0.1 - - [01/Dec/2025:20:55:36] "GET /admin/login.php  HTTP/1.1"404',
    '172.16.0.10 - - [01/Dec/2025:21:00:00] "GET /vuln/board.php  HTTP/1.1"200',
    '10.10.10.55 - - [01/Dec/2025:21:05:12] "POST /admin_shell.php  HTTP/1.1"500'
]

attack_db = {#Dictionary
    "admin": "관리자 페이지 접근 시도",
    "script": "XSS 공격(스크립트 삽입) 시도",
    "union": "SQL 인젝션 공격 시도",
    "../": "디렉토리 트래버설(상위 폴더 접근) 시도"
}
for label in log_list:
    lab = label.split()
    print(lab)
```

```python
#한 줄이 들어가야 하는데 세 줄이 들어가니까 당연히 오류 발생.
for log in log_list: #Iterate through the list
    r_ip,r_time,r_url = parse_log(log) #Unpacking Conception
    target_url = r_url.lower()#delete hiding
    print(target_url)
    if 'admin' in r_url:
      print(f"ALERT! Admin Access Detected{r_ip}")
      print(f"      ->URL:{r_url}")

print("=== Scanning Complete")
```

```python
#12. 10 잠시. time을 빼 보자고 - 수정 완료
def parse_log(line):
  part = line.split()
  ip = part[0]
  time = "".join(part[3:5]).strip('[]')
  attack_url = part[5]
  return ip,time,attack_url#어쨋든 return을 해야 함

def save_to_blacklist(ip,attack_type,time):#이걸 기준으로 txt 출력
  #open blacklist.txt.file as append'a'
  #encoding = 'UTF-8- 한글 안 깨지게.
  with open("blacklist.txt","a",encoding='UTF-8') as f:
    #파일에 쓸 내용(끝에 \n을 붙여야 줄바꿈이 됨.)
    f.write(f"[차단]{time}|{ip}|{attack_type}\n")

log_list = [
    '192.168.0.1 - - [01/Dec/2025:20:55:36] "GET /admin/login.php  HTTP/1.1"404',
    '172.16.0.10 - - [01/Dec/2025:21:00:00] "GET /vuln/board.php  HTTP/1.1"200',
    '10.10.10.55 - - [01/Dec/2025:21:05:12] "POST /admin_shell.php  HTTP/1.1"500'
]

attack_db = {#Dictionary
    "admin": "관리자 페이지 접근 시도",
    "script": "XSS 공격(스크립트 삽입) 시도",
    "union": "SQL 인젝션 공격 시도",
    "../": "디렉토리 트래버설(상위 폴더 접근) 시도"
}
for log in log_list: #Iterate through the list
    r_ip,r_time,r_url = parse_log(log) #Unpacking Conception
    target_url = r_url.lower()#delete hiding
    for keyword,attack_msg in attack_db.items():
      if keyword in target_url:
       console.print(f"ALERT! [bold red]{attack_msg} [/bold red]{r_ip}")
       print(f"      ->URL:{r_url}")
      save_to_blacklist(r_ip,attack_msg,r_time)
    console.print("[bold red]Hacking attemption detected [/bold red]")
print("=== Scanning Complete")
```
